#### Pregunta de reflexión (obligatoria en Markdown):

¿Qué información se pierde o se distorsiona al forzar una fuente no estructurada/semi-estructurada dentro de una tabla rectangular?

**Respuesta:**

Al forzar una fuente no estructurada o semi-estructurada (como un JSON con campos anidados o listas) en una tabla rectangular, se pueden perder o distorsionar varios tipos de información:

1.  **Jerarquía y relaciones inherentes:** La estructura jerárquica original del JSON que representa relaciones padre-hijo se pierde. Se reemplaza por una estructura plana donde estas relaciones deben inferirse a través de convenciones de nombres de columnas (ej., `campo_padre.campo_hijo`) o claves primarias/foráneas si se normaliza en múltiples tablas. Esto puede dificultar la comprensión de las relaciones originales de los datos.
2.  **Contexto semántico:** Elementos que están naturalmente agrupados en el formato original pueden ser separados, perdiendo parte de su contexto. Por ejemplo, una lista de eventos dentro de un registro principal, cuando se aplanan, pueden parecer registros independientes, perdiendo la conexión directa con el registro principal del que provienen, a menos que se replique la información del padre.
3.  **Redundancia de datos:** Si una entidad principal tiene múltiples elementos anidados (ej., un usuario con múltiples direcciones), al aplanar, la información del usuario se replicará para cada dirección. Esto introduce redundancia y aumenta el tamaño de la tabla, además de potencialmente dificultar las actualizaciones.
4.  **Información sobre la ausencia de datos:** Un campo que no existe en algunos registros JSON podría representarse como `NaN` (Not a Number) o nulo en la tabla rectangular. Esto podría interpretarse como un dato faltante cuando en realidad, en la fuente original, el campo simplemente no aplicaba o no estaba presente, lo que es una diferencia semántica importante.
5.  **Flexibilidad y granularidad:** Las fuentes semi-estructuradas permiten una gran flexibilidad en la adición de nuevos campos o la variación de la estructura de los registros. Al aplanar a un esquema fijo de tabla, se pierde esta flexibilidad, y cualquier nueva variación en la estructura del JSON requeriría una alteración del esquema de la tabla o un proceso de aplanamiento más complejo.
6.  **Complejidad en consultas:** Ciertas consultas que eran sencillas en el formato original (ej., “dame todos los eventos de este usuario”) pueden volverse más complejas en el formato aplanado, requiriendo uniones o agrupaciones para reconstruir la información original.

En resumen, mientras que el aplanamiento hace que los datos sean accesibles para herramientas y modelos que esperan formatos rectangulares, es crucial ser consciente de las posibles pérdidas y distorsiones semánticas y estructurales que esto implica.

### Problema: Formatos no validos (Fecha y hora)

Se fuerzan las fechas al formato correcto usando la coerción de pandas (errors="coerce"). Esto convierte la basura en valores nulos que luego son descartados. Decidimos hacer esto ya que los formatos de fecha y hora mixtos o inválidos impiden ordenar correctamente los eventos en el tiempo. Esto dificulta el análisis temporal y puede provocar errores graves al calcular tendencias o identificar los horarios reales de mayor tráfico.

Ademas, los identificadores estructurales que marcan el "cuándo" y el "dónde" no se pueden deducir si vienen dañados. Se deberian de excluir.

```python
# Tipos y formatos
trabajo["timestamp"] = pd.to_datetime(trabajo["timestamp"], errors="coerce")
for columna in ["conteo_vehiculos", "temperatura_c", "lat", "lon"]:
    trabajo[columna] = pd.to_numeric(trabajo[columna], errors="coerce")
```

### Problema: Categorías inconsistentes

Limpiamos los textos eliminando espacios sobrantes, aplicando mayúsculas iniciales (.str.title()) y utilizando diccionarios manuales para mapear sinónimos a una sola categoría (por ejemplo, homologar "Lluvioso" o "Nubes" a "Lluvia" y "Nublado"). 

El criterio utilizado para decidir qué categorías son la misma se basó en agrupar aquellos términos que semánticamente describen el mismo fenómeno meteorológico pero que fueron ingresados de forma diferente debido a errores de captura manuales o fuentes inconsistentes. Si estos no se agrupan, una misma condición climática se contabilizaría como categorías distintas. A nivel de negocio, esto es un riesgo porque fragmenta y afecta los análisis agrupados, distorsiona las visualizaciones y perjudica cualquier modelo futuro que pretenda utilizar el clima como una variable predictora del nivel de tráfico.

```python
# Estandarización de categorías

for columna in ["ubicacion", "tipo_via", "condicion_clima"]:
    trabajo[columna] = (trabajo[columna].astype("string").str.strip().str.replace(r"\s+", " ", regex=True).str.title())
trabajo["condicion_clima"] = trabajo["condicion_clima"].replace({
    "Lluvioso": "Lluvia", "Soleada": "Soleado", "Sol": "Soleado", "Nubes": "Nublado"
})
```

### Problema: Duplicados exactos y Duplicados de eventos de negocio

Para detectar los duplicados reales de nuestro problema, se utilizan las columnas sensor_id y timestamp de manera conjunta. Se elige esta combinación porque define de forma única e irrepetible un evento en el mundo físico. Es lógicamente imposible que un mismo sensor de movilidad registre dos realidades de tráfico distintas en el mismo segundo exacto. 

- Duplicados exactos: Eliminamos las filas redundantes con .drop_duplicates() para no inflar los cálculos descriptivos.  

- Duplicados de evento de negocio: Se elimina la duplicidad manteniendo la primera observación válida por cada llave lógica (sensor_id + timestamp).

Un .drop_duplicates() ingenuo no detectaría este error porque las filas enteras no son 100% iguales. Si no se filtran utilizando la llave de negocio, un mismo registro puede pesar varias veces en los análisis. Esto infla artificialmente los promedios de tráfico en ciertos horarios, lo que derivaría en una decisión errónea a la hora de priorizar la semaforización en la ciudad.


```python
# Duplicados: se conserva una observación por evento sensor-fecha-hora 
trabajo = trabajo.drop_duplicates().copy()
trabajo = trabajo.drop_duplicates(subset=["sensor_id", "timestamp"], keep="first").copy()
```

### Problema: Georreferenciación inválida

Coordenadas invertidas: Por errores de captura o del sistema, a veces el valor numérico de la longitud se guarda en la columna de la latitud y viceversa. El código corrige esto automáticamente intercambiando los valores cuando el valor absoluto de la latitud es mayor a 20 y el de la longitud es menor a 20.

Tambien definimos un rango geografico para limitar los valores extremos. Si se hace un intercambio (swap) de latitud y longitud sin una regla estricta, se podrían generar "falsos positivos" geográficos. Por eso, el código mitiga este riesgo exigiendo primero que los valores absolutos superen el umbral lógico de 20 grados para confirmar que es un error humano de digitación, y luego pasa el filtro de la caja geográfica de Medellín para mayor seguridad.


```python
# Validación geográfica aproximada del área metropolitana de Medellín.
# Si lat/lon están invertidas, se corrigen solo cuando el patrón es inequívoco.
coordenadas_invertidas = (trabajo["lat"].abs() > 20) & (trabajo["lon"].abs() < 20)
trabajo.loc[coordenadas_invertidas, ["lat", "lon"]] = trabajo.loc[coordenadas_invertidas, ["lon", "lat"]].to_numpy()
medellin_valida = trabajo["lat"].between(6.15, 6.35) & trabajo["lon"].between(-75.70, -75.50)
trabajo.loc[~medellin_valida, ["lat", "lon"]] = np.nan
```

### Problema: Conteos o temperaturas fuera de rango

Filtramos lógicamente los datos operativos. Definimos que los vehículos deben estar en el rango de 0 a 5,000 (ya que es un conteo por intervalo, no diario) y la temperatura de -10 a 45 grados Celsius. Todo lo que escape de allí se vuelve nulo y se imputa con la mediana.

```python
# Valores imposibles se convierten en faltantes; el límite de 5.000 vehículos
# corresponde a un conteo operativo por intervalo, no a un acumulado diario.
trabajo.loc[~trabajo["conteo_vehiculos"].between(0, 5000), "conteo_vehiculos"] = np.nan
trabajo.loc[~trabajo["temperatura_c"].between(-10, 45), "temperatura_c"] = np.nan
```

### Problema: Valores faltantes

- Valores numéricos (conteo_vehiculos, temperatura_c): Debemos imputar con la mediana. Al observar las estadísticas descriptivas de los datos contaminados: la columna de conteo de vehículos tiene valores atípicos extremos, con un máximo imposible de 99,999 y un mínimo de -5. Esto jala la media a un valor distorsionado de 328.03, mientras que la mediana se sitúa de forma realista en 19.0. La mediana es robusta ante estos valores atípicos.

- Variables categóricas (ubicacion, tipo_via, condicion_clima): Imputamos con la moda, rellenando los vacíos con la categoría más frecuente. Esto se debe a que las variables categoricas no tienen ni media ni mediana. Al llenar con la moda, continuamos el patron de repeticion.

- Identificadores estructurales (timestamp, lat, lon): Se debe eliminar la fila. Inventar fechas o coordenadas sin evidencia suficiente distorsiona la realidad y no aporta al análisis espacial temporal.


```python
# Imputación
for columna in ["conteo_vehiculos", "temperatura_c"]:
    trabajo[columna] = trabajo[columna].fillna(trabajo[columna].median())
for columna in ["ubicacion", "tipo_via", "condicion_clima"]:
    trabajo[columna] = trabajo[columna].fillna(trabajo[columna].mode().iloc[0])


# Fechas o coordenadas sin evidencia suficiente no se inventan: se excluyen del análisis.
df_trabajo = trabajo.dropna(subset=["timestamp", "lat", "lon"]).copy()

print(f"Filas contaminadas: {len(df_contaminado):,}")
print(f"Filas disponibles después de limpiar: {len(df_trabajo):,}")
display(df_trabajo.head())
```

#### Analisis Cuantitativo

Medidas de tendencia central y dispersión para al menos 2 variables continuas y 1 discreta, con la medida correcta según su distribución (¿usarían media o mediana? ¿por qué?). 

La siguiente tabla no solo reporta el valor de cada medida, sino también el **coeficiente de
asimetría (skewness)** de cada variable, que es el criterio usado para decidir entre
media y mediana.

**¿Por qué usamos este criterio?** La media es sensible a valores extremos: un solo dato muy
alto o muy bajo puede desplazarla lejos del centro real de los datos, mientras que la mediana es
resistente a outliers y refleja mejor el valor "típico" cuando la distribución está sesgada.


```python
# --- Tarea 4, punto 1: Cuantitativa ---
# Medidas de tendencia central y dispersión sobre los datos ya limpios (df_trabajo, Tarea 3).
# Variables continuas: temperatura_c, lat  |  Variable discreta: conteo_vehiculos

variables_continuas = ["temperatura_c", "lat"]
variable_discreta = "conteo_vehiculos"
variables_cuantitativas = variables_continuas + [variable_discreta]

def describir(serie):
    """Calcula medidas de tendencia central, dispersión y forma de una variable numérica."""
    q1, q3 = serie.quantile([0.25, 0.75])
    media = serie.mean()
    return pd.Series({
        "n": serie.count(),
        "media": media,
        "mediana": serie.median(),
        "desv_estandar": serie.std(),
        "coef_variacion": (serie.std() / media) if media else np.nan,
        "q1": q1,
        "q3": q3,
        "rango_intercuartilico": q3 - q1,
        "minimo": serie.min(),
        "maximo": serie.max(),
        "asimetria": serie.skew(),
    })

resumen_cuantitativo = pd.DataFrame(
    {col: describir(df_trabajo[col]) for col in variables_cuantitativas}
).T
resumen_cuantitativo["tipo"] = ["continua", "continua", "discreta"]

# Regla de decisión: |asimetría| < 0.5 -> distribución aprox. simétrica,
# por lo que la media es representativa; |asimetría| >= 0.5 -> distribución sesgada y/o con
# valores extremos, por lo que la mediana (robusta a outliers) es la medida preferida.
resumen_cuantitativo["medida_recomendada"] = np.where(
    resumen_cuantitativo["asimetria"].abs() < 0.5, "Media", "Mediana"
)

display(resumen_cuantitativo.round(3))
```

- temperatura_c — el caso más limpio. Con asimetría de -0.087 (prácticamente 0), la distribución es casi perfectamente simétrica. Media (23.07 °C) y mediana (23.10 °C) casi coinciden. Aquí la elección de la media está totalmente justificada: no hay sesgo que la distorsione, y además la media aprovecha toda la información de la muestra (mientras que la mediana descarta la magnitud de cada dato, solo usa el orden).

- conteo_vehiculos — el caso de sesgo real. Asimetría de 0.622 cae en el rango "moderadamente sesgada". Esto se confirma con la brecha entre media (22.45) y mediana (19.00): la media está siendo "jalada" hacia arriba por observaciones de tráfico alto (hasta 54 vehículos) en una variable que va de 0 a 54. Este es exactamente el patrón de un sesgo positivo con cola derecha. Aquí la mediana es la medida correcta — es consistente, además, con la decisión que ya tomaron en la Tarea 3 de imputar los faltantes de esta misma columna con la mediana precisamente por su robustez ante valores extremos.

- lat — Numéricamente cae en zona de sesgo moderado (0.691), pero la diferencia real entre media (6.239) y mediana (6.232) es de apenas 0.007 grados — insignificante en términos prácticos. Esto pasa porque lat tiene un rango naturalmente muy angosto (la caja de Medellín: 6.15–6.35), así que hasta una asimetría pequeña en términos absolutos puede producir un coeficiente de asimetría "moderado" sin que eso implique una distorsión real del centro de los datos. Basicamente, el rango es demasiado pequeño como para observar diferencias altas entre la media y la mediana, pero al agregar mas valores al conjunto de datos, la asimetria de esta variable se haria notar mas. La regla numérica recomienda mediana, pero el impacto práctico de usar media vs. mediana aquí es prácticamente nulo, a diferencia de conteo_vehiculos donde sí importa.

#### Analisis Cualitativo


Tablas de frecuencia/proporción para al menos 2 variables categóricas,
   incluyendo el cruce (tabla de contingencia) entre una variable categórica y la variable
   ordinal o de resultado de su conjunto de datos.


Este conjunto de datos no contiene variables ordinales, por lo que discretizamos la variable conteo_vehiculos para poder cumplir con el objetivo.

```python
# --- Tarea 4, punto 2: Cualitativa ---
# Tablas de frecuencia/proporción para variables categóricas + tabla de contingencia
# entre una categórica y la variable ordinal (nivel_trafico) sobre df_trabajo (Tarea 3).

variables_categoricas = ["ubicacion", "tipo_via", "condicion_clima"]

def tabla_frecuencia(serie):
    """Frecuencia absoluta y proporción de cada categoría."""
    tabla = serie.value_counts().to_frame("frecuencia")
    tabla["proporcion"] = (tabla["frecuencia"] / len(serie)).round(3)
    return tabla

for columna in variables_categoricas:
    print(f"Frecuencia y proporción — {columna}:")
    display(tabla_frecuencia(df_trabajo[columna]))

# Variable ordinal/resultado: se discretiza conteo_vehiculos en 3 niveles (tercios de la
# distribución) para poder cruzarla con las variables categóricas.
df_trabajo["nivel_trafico"] = pd.qcut(
    df_trabajo["conteo_vehiculos"], q=3, labels=["Bajo", "Medio", "Alto"], duplicates="drop"
)

print("Tabla de contingencia (conteo absoluto) — tipo_via x nivel_trafico:")
contingencia_conteo = pd.crosstab(df_trabajo["tipo_via"], df_trabajo["nivel_trafico"])
display(contingencia_conteo)

print("Tabla de contingencia (proporción por fila) — tipo_via x nivel_trafico:")
contingencia_proporcion = pd.crosstab(
    df_trabajo["tipo_via"], df_trabajo["nivel_trafico"], normalize="index"
).round(3)
display(contingencia_proporcion)
```


- Las tablas de **frecuencia/proporción** de ubicacion, tipo_via y condicion_clima
  muestran qué tan representado está cada corredor, tipo de vía y condición climática dentro
  de los datos ya limpios; esto es clave para saber si alguna categoría tiene pocas
  observaciones y, por tanto, sus estadísticas son menos confiables.

- La **tabla de contingencia en conteos absolutos** (tipo_via x nivel_trafico) muestra
  cuántos registros caen en cada combinación, pero al tener tipos de vía con distinto número
  total de observaciones, comparar los conteos crudos entre filas puede ser engañoso.

- Por eso normalizamos por fila (normalize="index"): así cada fila de tipo_via suma 1 y
  el valor representa la **proporción de registros de ese tipo de vía** que cae en cada nivel
  de tráfico (Bajo/Medio/Alto). Esto permite comparar directamente, por ejemplo, si las
  autopistas concentran una mayor proporción de tráfico "Alto" que las vías secundarias,
  independientemente de cuántos sensores tenga cada tipo de vía.
